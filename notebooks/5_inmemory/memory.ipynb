```python
from langchain_core.messages import HumanMessage, AIMessage
from langchain_core.prompts import ChatPromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
llm=ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0
)
print("Gemini Successfuly Initialized")
```

```python
conversational_history=[]
print("Conversational Memory Initialised")
```

```python
def add_to_memory(question, answer):
    conversational_history.append(
        HumanMessage(content=question)
    )

    conversational_history.append(
        AIMessage(content=answer)
    )
```

```python
def show_memory():
    for message in conversational_history:
        if isinstance(message, HumanMessage):
            print("User:", message.content)
        elif isinstance(message, AIMessage):
            print("AI:", message.content)
```

```python
add_to_memory(
    "What is this video about?",
    "This video explains how artificial intelligence is changing the technology industry."
)

add_to_memory(
    "Who is the speaker?",
    "The speaker is discussing recent developments in artificial intelligence."
)

show_memory()
```

```python
def format_memory():
    history=""

    for message in conversational_history:
        if isinstance(message, HumanMessage):
            history += f"User: {message.content}\n"
        elif isinstance(message, AIMessage):
            history += f"Assistant: {message.content}\n"
    
    return history
```

```python
print(format_memory())
```

```python
memory_prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """You are a helpful assistant.

Use the conversation history to understand the user's current question,
especially when the user asks a follow-up question.

Conversation history:
{history}

Answer the current question clearly and naturally.
"""
    ),
    ("human", "{question}")
])
```

```python
memory_chain = memory_prompt|llm
print("Memory Chain Created")
```

```python
question = "What is artificial intelligence?"

response = memory_chain.invoke({
    "history": format_memory(),
    "question": question
})

answer = response.content

print("AI:", answer)

add_to_memory(question, answer)
```

```python
question = "Why is it important?"

response = memory_chain.invoke({
    "history": format_memory(),
    "question": question
})

answer = response.content

print("AI:", answer)

add_to_memory(question, answer)
```

```python
question = "can you explain it simpler terms?"

response = memory_chain.invoke({
    "history": format_memory(),
    "question": question
})

answer = response.content

print("AI:", answer)

add_to_memory(question, answer)
```

```python
show_memory()
```

```python
def clear_memory():
    conversational_history.clear()
    print("Memory cleared.")
```

```python
clear_memory()
```

```python
show_memory()
```
